números , integer (int)  153, -23, 0

van en llaves.
`{'a','b','c','d','e'}`

### Conversiones
Cuando queremos convertir un dato a otro se le llama casting
hay dos tipos implicitas y explicitas
- Implicitas: Python convierte el tipo de datos en otro tipo de dato automaticamente
- Explicita: el usuario dee de convertirlo en un tipo de dato manualmente.


```python
# Conversion implicita
num1 = 20
num2 = 30.5
print(type(num1))
num1 = num1 + num2
print(num1)
print(type(num1))
print(type(num2))
```

```python
# Conversion explicita
num1 = 5.8
print(num1)
print(type(num1))

num2 = int(num1)
print(num2)
print(type(num2))

edad = int(input("Escribe tu edad"))
print("Tu edad es ")
print(edad)
print(type(edad))

print("edad al doble")
print(edad + edad)
print(type(edad))

# Esto ya no es posible ya que al concatenar no sabe si sumar o tomarlo como str
# Esto se resolverá más adelante
# print("Tu edad es " + edad)
```

### Operadores matemáticos


```python
x = 6
y = 2
z = 7

print(f"{x} mas {y} es igual a {x + y}")
print(f"{x} mas {y} es igual a {x - y}")
print(f"{x} mas {y} es igual a {x * y}")
print(f"{x} mas {y} es igual a {x / y}")

# División al piso, toma en cuenta solo el valor entero
print(f"{z} dividido al piso de {y} es igual a {z // y}")
# módulo: el residuo de la división
print(f"{z} modulo de {y} es igual a {z % y}")
print(f"{x} elevado a {y} es igual a {x ** y}")
print(f"La raíz cuadrada de {x} es igual a {x ** 0.5}")

```

El resultado de la suma de al menos un número de punto flotante con otro integers o floats, será siempre un float


Dada una variable llamada num1, con un valor de 13.87:

`num1 = 13.87`

Las siguientes líneas en Python, arrojarían el mismo resultado:

`print(round(num1))`

`print(int(num1))`

Respuesta: Falso, round redonde al entero más próximo (14), mientras que int conserva la parte entera del valor (13), por lo que los resultados no seran iguales.
